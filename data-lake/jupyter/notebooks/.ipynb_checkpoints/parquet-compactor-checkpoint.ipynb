# Parquet Files Compactor

Компактор объединяет маленькие Parquet файлы (созданные из 72 партиций Kafka) в файлы оптимального размера 256-512 МБ.

## Как работает:
1. Читает все маленькие файлы из исходной партиции
2. Объединяет их в память
3. Записывает в оптимальное количество файлов нужного размера
4. Сохраняет в новую директорию `topics-compacted/`

````python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
import math

# Создание Spark сессии
spark = SparkSession.builder \
    .appName("Parquet-Compactor") \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")

print(f"Spark version: {spark.version}")
print(f"Spark UI: http://localhost:4040")
````

````python
# Конфигурация компактора
SOURCE_BUCKET = "datalake"
SOURCE_PREFIX = "topics/order-events"  # Исходная директория
TARGET_PREFIX = "topics-compacted/order-events"  # Целевая директория

# Целевой размер файлов в МБ
TARGET_FILE_SIZE_MB = 384  # Среднее между 256-512 МБ

# Compression ratio для Parquet (примерно)
# В памяти данные занимают больше, чем на диске
COMPRESSION_RATIO = 0.15  # Parquet с snappy дает ~15% от размера в памяти

print(f"Source: s3a://{SOURCE_BUCKET}/{SOURCE_PREFIX}/")
print(f"Target: s3a://{SOURCE_BUCKET}/{TARGET_PREFIX}/")
print(f"Target file size: {TARGET_FILE_SIZE_MB} MB")
````

````python
def analyze_partition(partition_path):
    """
    Анализирует партицию и возвращает статистику
    """
    df = spark.read.parquet(partition_path)
    
    # Подсчет количества записей
    row_count = df.count()
    
    # Оценка размера в памяти (примерно)
    # Предполагаем ~1KB на запись для order-events
    estimated_size_mb = row_count * 1024 / (1024 * 1024)
    
    # Оценка размера на диске после сжатия
    estimated_disk_size_mb = estimated_size_mb * COMPRESSION_RATIO
    
    # Оптимальное количество файлов
    optimal_file_count = max(1, math.ceil(estimated_disk_size_mb / TARGET_FILE_SIZE_MB))
    
    return {
        'row_count': row_count,
        'estimated_memory_mb': estimated_size_mb,
        'estimated_disk_mb': estimated_disk_size_mb,
        'optimal_file_count': optimal_file_count
    }

print("Function analyze_partition() created")
````

````python
def compact_partition(source_path, target_path, num_files=None, preserve_partitions=True):
    """
    Компактирует файлы из source_path в target_path
    
    Args:
        source_path: путь к исходным файлам (s3a://...)
        target_path: путь для сохранения (s3a://...)
        num_files: количество выходных файлов (если None, будет рассчитано автоматически)
        preserve_partitions: сохранять ли партиционирование по calc_id/dt/hour
    """
    print(f"\n{'='*70}")
    print(f"Starting compaction")
    print(f"Source: {source_path}")
    print(f"Target: {target_path}")
    print(f"{'='*70}\n")
    
    # Читаем данные
    print("Reading source files...")
    df = spark.read.parquet(source_path)
    
    # Анализ
    row_count = df.count()
    estimated_size_mb = row_count * 1024 / (1024 * 1024)
    estimated_disk_mb = estimated_size_mb * COMPRESSION_RATIO
    
    print(f"Total rows: {row_count:,}")
    print(f"Estimated memory size: {estimated_size_mb:.2f} MB")
    print(f"Estimated disk size: {estimated_disk_mb:.2f} MB")
    
    # Определяем количество файлов
    if num_files is None:
        num_files = max(1, math.ceil(estimated_disk_mb / TARGET_FILE_SIZE_MB))
    
    print(f"Target number of files: {num_files}")
    print(f"Expected file size: {estimated_disk_mb / num_files:.2f} MB each")
    
    # Уменьшаем количество партиций
    print(f"\nRepartitioning to {num_files} partitions...")
    df_compacted = df.coalesce(num_files)
    
    # Записываем результат
    print("Writing compacted files...")
    
    writer = df_compacted.write.mode("overwrite")
    
    # Настройки Parquet
    writer = writer \
        .option("compression", "snappy") \
        .option("parquet.block.size", 16777216) \
        .option("parquet.page.size", 524288)
    
    # Партиционирование
    if preserve_partitions and 'calc_id' in df.columns:
        print("Preserving partitioning by calc_id, dt, hour")
        writer = writer.partitionBy("calc_id", "dt", "hour")
    
    writer.parquet(target_path)
    
    print(f"\n{'='*70}")
    print(f"Compaction completed!")
    print(f"Files saved to: {target_path}")
    print(f"{'='*70}")

print("Function compact_partition() created")
````

## Использование

### Вариант 1: Анализ перед компактацией

````python
# Анализируем текущие данные
source_path = f"s3a://{SOURCE_BUCKET}/{SOURCE_PREFIX}/"
stats = analyze_partition(source_path)

print("\nPartition Statistics:")
print(f"  Rows: {stats['row_count']:,}")
print(f"  Estimated memory size: {stats['estimated_memory_mb']:.2f} MB")
print(f"  Estimated disk size: {stats['estimated_disk_mb']:.2f} MB")
print(f"  Recommended files: {stats['optimal_file_count']}")
print(f"  Expected file size: {stats['estimated_disk_mb'] / stats['optimal_file_count']:.2f} MB each")
````

### Вариант 2: Компактация всех данных

````python
# Компактация всех данных
source_path = f"s3a://{SOURCE_BUCKET}/{SOURCE_PREFIX}/"
target_path = f"s3a://{SOURCE_BUCKET}/{TARGET_PREFIX}/"

# Автоматическое определение количества файлов
compact_partition(source_path, target_path, preserve_partitions=True)
````

### Вариант 3: Компактация конкретной партиции

````python
# Компактация конкретной партиции (calc_id/dt/hour)
calc_id = "20251224-180000"
dt = "2025-12-24"
hour = "18"

source_path = f"s3a://{SOURCE_BUCKET}/{SOURCE_PREFIX}/calc_id={calc_id}/dt={dt}/hour={hour}/"
target_path = f"s3a://{SOURCE_BUCKET}/{TARGET_PREFIX}/calc_id={calc_id}/dt={dt}/hour={hour}/"

# Указываем количество файлов вручную (например, 2 файла)
compact_partition(source_path, target_path, num_files=2, preserve_partitions=False)
````

## Проверка результатов

````python
# Проверяем результаты компактации
import s3fs

s3 = s3fs.S3FileSystem(
    key='minioadmin',
    secret='minioadmin',
    client_kwargs={'endpoint_url': 'http://minio:9000'}
)

# Исходные файлы
source_files = s3.glob(f'{SOURCE_BUCKET}/{SOURCE_PREFIX}/**/*.parquet')
print(f"\nSource files: {len(source_files)}")

source_size_mb = sum([s3.size(f) for f in source_files]) / (1024 * 1024)
print(f"Total source size: {source_size_mb:.2f} MB")
print(f"Average file size: {source_size_mb / len(source_files):.2f} MB")

# Компактированные файлы
target_files = s3.glob(f'{SOURCE_BUCKET}/{TARGET_PREFIX}/**/*.parquet')
if target_files:
    print(f"\nCompacted files: {len(target_files)}")
    
    target_size_mb = sum([s3.size(f) for f in target_files]) / (1024 * 1024)
    print(f"Total compacted size: {target_size_mb:.2f} MB")
    print(f"Average file size: {target_size_mb / len(target_files):.2f} MB")
    
    print(f"\nReduction: {len(source_files)} files -> {len(target_files)} files")
    print(f"Reduction factor: {len(source_files) / len(target_files):.1f}x")
else:
    print("\nNo compacted files found yet. Run compaction first.")
````

## Автоматизация (для продакшена)

Для автоматической компактации можно:

1. **Создать cron job**, который запускает этот notebook каждый час/день
2. **Использовать Airflow DAG** для оркестрации
3. **Настроить Kafka Connect** для записи сразу в оптимальном формате (но это требует изменения топологии)

Пример запуска через papermill:
```bash
papermill parquet-compactor.ipynb output.ipynb \
  -p SOURCE_PREFIX "topics/order-events" \
  -p TARGET_PREFIX "topics-compacted/order-events"
```
