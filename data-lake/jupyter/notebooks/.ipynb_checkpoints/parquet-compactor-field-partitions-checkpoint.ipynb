# Parquet Compactor для Field-Partitioned данных

Компактор для данных, разбитых по полям (FieldPartitioner).

## Проблема:
- 72 партиции Kafka × N значений prtNum × M значений billing_country = тысячи мелких файлов
- Каждый файл ~10 МБ вместо оптимальных 256-512 МБ

## Решение:
Компактация внутри каждой комбинации (prtNum, billing_country)

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
import math

spark = SparkSession.builder \
    .appName("Field-Partition-Compactor") \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")

print(f"Spark version: {spark.version}")
print(f"Spark UI: http://localhost:4040")
```

```python
# Конфигурация
SOURCE_BUCKET = "datalake"
SOURCE_PREFIX = "topics/order-events"  # Данные с FieldPartitioner
TARGET_PREFIX = "topics-compacted/order-events-by-fields"

TARGET_FILE_SIZE_MB = 384  # Целевой размер файла
COMPRESSION_RATIO = 0.15

print(f"Source: s3a://{SOURCE_BUCKET}/{SOURCE_PREFIX}/")
print(f"Target: s3a://{SOURCE_BUCKET}/{TARGET_PREFIX}/")
```

```python
# Анализ: сколько уникальных комбинаций prtNum+billing_country?
source_path = f"s3a://{SOURCE_BUCKET}/{SOURCE_PREFIX}/"
df = spark.read.parquet(source_path)

print("Analyzing field partitions...\n")

# Общая статистика
total_rows = df.count()
print(f"Total rows: {total_rows:,}")

# Если есть поля для партиционирования
if 'prtNum' in df.columns and 'billing_country' in df.columns:
    partition_stats = df.groupBy('prtNum', 'billing_country').count() \
        .orderBy('count', ascending=False)
    
    num_partitions = partition_stats.count()
    print(f"\nUnique (prtNum, billing_country) combinations: {num_partitions}")
    
    print("\nTop 10 partitions by size:")
    partition_stats.show(10)
    
    # Статистика распределения
    stats_df = partition_stats.agg(
        min('count').alias('min_rows'),
        max('count').alias('max_rows'),
        avg('count').alias('avg_rows'),
        stddev('count').alias('stddev_rows')
    ).collect()[0]
    
    print(f"\nPartition size distribution:")
    print(f"  Min rows: {stats_df['min_rows']:,}")
    print(f"  Max rows: {stats_df['max_rows']:,}")
    print(f"  Avg rows: {stats_df['avg_rows']:,.0f}")
    print(f"  Stddev: {stats_df['stddev_rows']:,.0f}")
else:
    print("\nNo prtNum or billing_country fields found in data")
    print("Available columns:", df.columns)
```

## Стратегия компактации

Есть два подхода:

### Вариант A: Сохранить field-партиционирование
- Компактировать файлы внутри каждой комбинации (prtNum, billing_country)
- Каждая комбинация → 1 файл (или несколько, если данных много)
- Сохраняет структуру директорий: `prtNum=X/billing_country=Y/`

### Вариант B: Убрать field-партиционирование
- Объединить все в файлы по 256-512 МБ без разделения по полям
- Проще для анализа, меньше файлов
- Но теряется возможность быстрого фильтра по prtNum/billing_country

```python
# ВАРИАНТ A: Компактация с сохранением field-партиционирования

def compact_with_field_partitions(source_path, target_path):
    """
    Компактация с сохранением партиционирования по prtNum и billing_country.
    Внутри каждой комбинации объединяет 72 маленьких файла (по числу Kafka партиций) в 1-2 файла.
    """
    print(f"Reading data from {source_path}...")
    df = spark.read.parquet(source_path)
    
    total_rows = df.count()
    print(f"Total rows: {total_rows:,}")
    
    # Группируем по партициям для расчета размера
    if 'prtNum' in df.columns and 'billing_country' in df.columns:
        partition_stats = df.groupBy('prtNum', 'billing_country').count().collect()
        
        total_partitions = len(partition_stats)
        avg_rows_per_partition = total_rows / total_partitions
        
        print(f"\nPartitions: {total_partitions}")
        print(f"Avg rows per partition: {avg_rows_per_partition:,.0f}")
        
        # Для каждой партиции будет создан 1 файл (или больше, если данных много)
        # Используем repartition с числом партиций = количество уникальных комбинаций
        # Это гарантирует, что каждая комбинация попадет в свой файл
        
        print(f"\nRepartitioning by fields...")
        df_repartitioned = df.repartition('prtNum', 'billing_country')
        
        print(f"Writing compacted files to {target_path}...")
        df_repartitioned.write.mode("overwrite") \
            .option("compression", "snappy") \
            .option("parquet.block.size", 16777216) \
            .option("parquet.page.size", 524288) \
            .partitionBy("prtNum", "billing_country") \
            .parquet(target_path)
        
        print("\nCompaction completed!")
        print(f"Each (prtNum, billing_country) combination is now in 1 file instead of 72")
    else:
        print("ERROR: prtNum or billing_country not found in data")

# Запуск
source_path = f"s3a://{SOURCE_BUCKET}/{SOURCE_PREFIX}/"
target_path = f"s3a://{SOURCE_BUCKET}/{TARGET_PREFIX}/"

compact_with_field_partitions(source_path, target_path)
```

```python
# ВАРИАНТ B: Компактация БЕЗ field-партиционирования
# (убираем разделение по prtNum и billing_country, группируем только по времени)

def compact_remove_field_partitions(source_path, target_path, num_files=None):
    """
    Компактация с удалением field-партиционирования.
    Все данные объединяются в N файлов оптимального размера.
    """
    print(f"Reading data from {source_path}...")
    df = spark.read.parquet(source_path)
    
    total_rows = df.count()
    estimated_size_mb = total_rows * 1024 / (1024 * 1024)
    estimated_disk_mb = estimated_size_mb * COMPRESSION_RATIO
    
    print(f"Total rows: {total_rows:,}")
    print(f"Estimated disk size: {estimated_disk_mb:.2f} MB")
    
    if num_files is None:
        num_files = max(1, math.ceil(estimated_disk_mb / TARGET_FILE_SIZE_MB))
    
    print(f"Target files: {num_files}")
    print(f"Expected file size: {estimated_disk_mb / num_files:.2f} MB each")
    
    print(f"\nCoalescing to {num_files} partitions...")
    df_compacted = df.coalesce(num_files)
    
    print(f"Writing compacted files to {target_path}...")
    
    # Можем сохранить партиционирование по времени, если нужно
    if 'calc_id' in df.columns:
        df_compacted.write.mode("overwrite") \
            .option("compression", "snappy") \
            .option("parquet.block.size", 16777216) \
            .option("parquet.page.size", 524288) \
            .partitionBy("calc_id", "dt", "hour") \
            .parquet(target_path)
    else:
        df_compacted.write.mode("overwrite") \
            .option("compression", "snappy") \
            .option("parquet.block.size", 16777216) \
            .option("parquet.page.size", 524288) \
            .parquet(target_path)
    
    print("\nCompaction completed!")
    print(f"Files reduced from thousands to {num_files}")

# Запуск (закомментирован, раскомментируйте для использования)
# source_path = f"s3a://{SOURCE_BUCKET}/{SOURCE_PREFIX}/"
# target_path = f"s3a://{SOURCE_BUCKET}/topics-compacted/order-events-no-field-partitions/"
# compact_remove_field_partitions(source_path, target_path)
```

## Проверка результатов

```python
import s3fs

s3 = s3fs.S3FileSystem(
    key='minioadmin',
    secret='minioadmin',
    client_kwargs={'endpoint_url': 'http://minio:9000'}
)

# Исходные файлы
source_files = s3.glob(f'{SOURCE_BUCKET}/{SOURCE_PREFIX}/**/*.parquet')
print(f"Source files: {len(source_files)}")
source_size_mb = sum([s3.size(f) for f in source_files]) / (1024 * 1024)
print(f"Total source size: {source_size_mb:.2f} MB")
print(f"Average source file: {source_size_mb / len(source_files):.2f} MB")

# Компактированные файлы
target_files = s3.glob(f'{SOURCE_BUCKET}/{TARGET_PREFIX}/**/*.parquet')
if target_files:
    print(f"\nCompacted files: {len(target_files)}")
    target_size_mb = sum([s3.size(f) for f in target_files]) / (1024 * 1024)
    print(f"Total compacted size: {target_size_mb:.2f} MB")
    print(f"Average compacted file: {target_size_mb / len(target_files):.2f} MB")
    
    print(f"\nReduction: {len(source_files)} files -> {len(target_files)} files")
    print(f"Reduction factor: {len(source_files) / len(target_files):.1f}x")
    
    # Показать размеры компактированных файлов
    print("\nCompacted file sizes:")
    for f in sorted(target_files)[:20]:  # Первые 20
        size_mb = s3.size(f) / (1024 * 1024)
        print(f"  {f.split('/')[-1]}: {size_mb:.2f} MB")
else:
    print("\nNo compacted files found. Run compaction first.")
```

## Рекомендации

### Когда использовать Вариант A (с сохранением field-партиций):
- Если часто фильтруете данные по `prtNum` или `billing_country`
- Если нужна предикатная pushdown оптимизация при чтении
- Если количество уникальных комбинаций не очень большое (< 1000)

### Когда использовать Вариант B (без field-партиций):
- Если читаете всю таблицу целиком (full scan)
- Если количество комбинаций очень большое (тысячи)
- Если хотите минимизировать общее количество файлов
- Если фильтрация по этим полям не критична

### Для продакшена:
1. Запускайте компактацию периодически (каждый час/день)
2. Используйте Airflow или cron для автоматизации
3. Мониторьте размеры файлов и корректируйте `TARGET_FILE_SIZE_MB`
4. Рассмотрите удаление исходных файлов после успешной компактации
